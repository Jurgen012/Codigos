**Calculamos el R2**

```python
#Calcular R^2 de (1)
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv("Advertising.csv")

x = df[['TV', 'radio', 'newspaper']]
y = df['sales']

lr = LinearRegression()
lr.fit(x,y)

r2 = lr.score(x,y)
r2
```

**Calculamos el arbol de decisión y sacamos su r2**

```python
from sklearn.tree import DecisionTreeRegressor

#arbol
tree = DecisionTreeRegressor(random_state=67)
tree.fit(x, y)
#r2
r2_tree = tree.score(x, y)
r2_tree
```

### Repetimos el proceso con un Train-TestSplit (50/50)

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.5, random_state=42)

# (1) Regresión lineal
lr2 = LinearRegression()
lr2.fit(x_train, y_train)

# (2) R² regresión lineal
r2_lrtest = lr2.score(x_test, y_test)
r2_lrtrain = lr2.score(x_train, y_train)
print(f"R² Regresión Lineal Test:      {r2_lrtest}")
print(f"R² Regresión Lineal Train:      {r2_lrtrain}")

#------------------------------------------------------------

# (3) Árbol de decisión
tree2 = DecisionTreeRegressor(random_state=42)
tree2.fit(x_train, y_train)

# (4) R² árbol de decisión
r2_treetest = tree2.score(x_test, y_test)
r2_treetrain = tree2.score(x_train,y_train)

print(f"R² Árbol de Decisión Test:     {r2_treetest}")
print(f"R² Árbol de Decisión Train: {r2_treetrain}")
```

# Actividad 1: **Arbol de Decisión (Estatura)**

```python
import pandas as pd

dataframe = pd.DataFrame({
    "Estatura": [1.73,1.62,1.73,1.69,1.75,1.80,1.70,1.75,1.66,1.64,1.62,1.85,1.75,1.66,1.70,1.64],
    "Edad": [21,19,21,19,20,21,22,21,19,21,19,21,20,20,19,19],
    "Genero": ["H","M","H","H","H","H","H","H","H","M","M","H","H","H","H","H"]
})
print(dataframe)

#promedio de estatura
promedio_estatura = dataframe["Estatura"].mean()
print("Promedio de estatura:", promedio_estatura)
```

```python
#promedio de estatura hombres mayor o igual 21 años
hombres_mayores_21 = dataframe[(dataframe["Genero"] == "H") & (dataframe["Edad"] >= 21)]
promedio_estatura_hombres_mayores_21 = hombres_mayores_21["Estatura"].mean()
print("Promedio de estatura de hombres mayor o iguales a 21 años:", promedio_estatura_hombres_mayores_21)

#promedio de estatura hombres menor a 21 años
hombres_menores_21 = dataframe[(dataframe["Genero"] == "H") & (dataframe["Edad"] < 21)]
promedio_estatura_hombres_menores_21 = hombres_menores_21["Estatura"].mean()
print("Promedio de estatura de hombres menores a 21 años:", promedio_estatura_hombres_menores_21)

#------------------------------------------------------------------------------------------------------

#promedio de estatura mujeres mayor o igual 21 años
mujeres_mayores_21 = dataframe[(dataframe["Genero"] == "M") & (dataframe["Edad"] >= 21)]
promedio_estatura_mujeres_mayores_21 = mujeres_mayores_21["Estatura"].mean()
print("Promedio de estatura de mujeres mayor o iguales a 21 años:", promedio_estatura_mujeres_mayores_21)

#promedio de estatura mujeres menor a 21 años
mujeres_menores_21 = dataframe[(dataframe["Genero"] == "M") & (dataframe["Edad"] < 21)]
promedio_estatura_mujeres_menores_21 = mujeres_menores_21["Estatura"].mean()
print("Promedio de estatura de mujeres menores a 21 años:", promedio_estatura_mujeres_menores_21)
```
