## Dataset
Motor Trend Car Road Tests (mtcars) contiene información de
32 autos con variables como cilindros, peso, potencia, etc.

## Objetivo
Aprender a construir y evaluar modelos de regresión lineal
variando tres elementos:

### **1.1 Realiza una regresión tomando 'mpg' como salida y eliminando la columna 'model'.**

````python
import pandas as pd

Motor = pd.read_excel("Motor Trend Car Road Tests.xlsx")
````

````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import r2_score

Motor = pd.read_excel("Motor Trend Car Road Tests.xlsx")

# ── EDA / Split ───────────────────────────────────────────
muestra = Motor.sample(frac=0.40, random_state=123)   # 40% train

Y = muestra["mpg"]
X = muestra.drop(["mpg", "model"], axis=1)

Xtest = Motor.drop(muestra.index).drop(["mpg", "model"], axis=1)
Ytest = Motor.drop(muestra.index)["mpg"]

# ── Escalado ──────────────────────────────────────────────
scaler = StandardScaler()
Xfinal      = scaler.fit_transform(X)       # fit solo en train
Xtest_final = scaler.transform(Xtest)       # solo transform en test

# ── 1. Regresión Lineal (OLS) ─────────────────────────────
modelo = LinearRegression()
modelo.fit(Xfinal, Y)

# R² train
Ypred_train = modelo.predict(Xfinal)
r2_train1 = r2_score(Y, Ypred_train)

# R² test
Ypred_test = modelo.predict(Xtest_final)
r2_test1 = r2_score(Ytest, Ypred_test)

print("=== OLS ===")
print(f"R² Train : {r2_train1:.4f}")
print(f"R² Test  : {r2_test1:.4f}")

# Betas con sus nombres
betas = pd.Series(modelo.coef_, index=X.columns)
print("\nBetas (coeficientes estandarizados):")
print(betas.sort_values())

# ── 2. Ridge con varios lambdas ───────────────────────────
lambdas = [0.01, 0.1, 1, 10, 50, 100, 500, 1000]

print("\n=== Ridge ===")
print(f"{'Lambda':>8}  {'R² Train':>10}  {'R² Test':>10}")

for lam in lambdas:
    ridge = Ridge(alpha=lam)
    ridge.fit(Xfinal, Y)
    rt  = r2_score(Y,     ridge.predict(Xfinal))
    rte = r2_score(Ytest, ridge.predict(Xtest_final))
    print(f"{lam:>8}  {rt:>10.4f}  {rte:>10.4f}")
````

### **2.1 Realiza una regresión tomando 'mpg' como salida y eliminando la columna 'model'. Crea columnas dummies para los factores 'cyl', 'gear' y 'carb'.**

````python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

Motor = pd.read_excel("Motor Trend Car Road Tests.xlsx")

# ── Dummies ───────────────────────────────────────────────
# drop_first=True evita la trampa de la variable dummy (multicolinealidad perfecta)
Motor = pd.get_dummies(Motor, columns=["cyl", "gear", "carb"], drop_first=True)

# ── EDA / Split ───────────────────────────────────────────
muestra = Motor.sample(frac=0.40, random_state=123)

Y = muestra["mpg"]
X = muestra.drop(["mpg", "model"], axis=1)

Xtest = Motor.drop(muestra.index).drop(["mpg", "model"], axis=1)
Ytest = Motor.drop(muestra.index)["mpg"]

# ── Escalado ──────────────────────────────────────────────
scaler = StandardScaler()
Xfinal      = scaler.fit_transform(X)
Xtest_final = scaler.transform(Xtest)

# ── Regresión OLS ─────────────────────────────────────────
modelo = LinearRegression()
modelo.fit(Xfinal, Y)

# ── R² Train ──────────────────────────────────────────────
r2_train3 = r2_score(Y, modelo.predict(Xfinal))

# ── R² Test ───────────────────────────────────────────────
r2_test3 = r2_score(Ytest, modelo.predict(Xtest_final))

print(f"R² Train : {r2_train3:.4f}")
print(f"R² Test  : {r2_test3:.4f}")

# ── Betas ─────────────────────────────────────────────────
betas = pd.Series(modelo.coef_, index=X.columns)
print("\nBetas (coeficientes estandarizados):")
print(betas.sort_values())
````

### **2.2 Repite el ejercicio anterior usando 'qsec' como salida.**

````python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

Motor = pd.read_excel("Motor Trend Car Road Tests.xlsx")

# ── Dummies ───────────────────────────────────────────────
Motor = pd.get_dummies(Motor, columns=["cyl", "gear", "carb"], drop_first=True)

# ── EDA / Split ───────────────────────────────────────────
muestra = Motor.sample(frac=0.40, random_state=123)

Y = muestra["qsec"]                                    # ← cambio
X = muestra.drop(["qsec", "model"], axis=1)            # ← cambio

Xtest = Motor.drop(muestra.index).drop(["qsec", "model"], axis=1)
Ytest = Motor.drop(muestra.index)["qsec"]              # ← cambio

# ── Escalado ──────────────────────────────────────────────
scaler = StandardScaler()
Xfinal      = scaler.fit_transform(X)
Xtest_final = scaler.transform(Xtest)

# ── Regresión OLS ─────────────────────────────────────────
modelo = LinearRegression()
modelo.fit(Xfinal, Y)

# ── R² Train ──────────────────────────────────────────────
r2_train4 = r2_score(Y, modelo.predict(Xfinal))

# ── R² Test ───────────────────────────────────────────────
r2_test4 = r2_score(Ytest, modelo.predict(Xtest_final))

print(f"R² Train : {r2_train4:.4f}")
print(f"R² Test  : {r2_test4:.4f}")

# ── Betas ─────────────────────────────────────────────────
betas = pd.Series(modelo.coef_, index=X.columns)
print("\nBetas (coeficientes estandarizados):")
print(betas.sort_values())
````

### **Comparar R^2 de los ejercicios 1.1 y 2.1**

````python
print(f"R² Train 1.1 (numéricos): {r2_train1:.4f}")
print(f"R² Train 2.1 (dummies)  : {r2_train2:.4f}")

print(f"R² Test 1.1 (numéricos): {r2_test1:.4f}")
print(f"R² Test 2.1 (dummies)  : {r2_test2:.4f}")
````

### Train
Ambos modelos ajustan bien en entrenamiento (~0.96–0.98),
lo cual es esperado con tan pocos datos (~13 observaciones).

### Test
Ambos R² Test son negativos, lo que significa que los modelos
son **peores que predecir la media** de `mpg`. Esto es sobreajuste
clásico: demasiados parámetros para muy pocas observaciones.

### ¿Por qué 2.1 es peor?
Las dummies sobre `cyl`, `gear` y `carb` añaden columnas extra,
aumentando aún más la relación parámetros/datos y agravando el sobreajuste.

## Conclusión
Con un dataset tan pequeño como mtcars (32 filas, 40% ≈ 13 train),
tratar las variables como **numéricas es preferible** a crear dummies.
Las dummies son útiles cuando hay suficientes datos para absorber
el incremento de dimensionalidad.

### **Comparar R^2 de los ejercicios 1.2 y 2.2**

````python

print(f"R² Train 1.2 (numéricos): {r2_train3:.4f}")
print(f"R² Train 2.2 (dummies)  : {r2_train4:.4f}")

print(f"R² Test 1.2 (numéricos): {r2_test3:.4f}")
print(f"R² Test 2.2 (dummies)  : {r2_test4:.4f}")
````

### Train
Ambos modelos memorizan perfectamente el train (R² = 1.0),
lo cual no es un logro sino una señal de sobreajuste total.

### Test
Aquí ocurre algo interesante e **inverso** a lo que pasó en 1.1 vs 2.1:

- En 1.1 vs 2.1 las dummies **empeoraban** el test (–1.12 → –8.47)
- En 1.2 vs 2.2 las dummies **mejoran** el test (–25,294 → –10.93)

### ¿Por qué las dummies ayudan con `qsec`?
El modelo numérico 1.2 trata `cyl`, `gear` y `carb` como
valores continuos, lo que genera coeficientes extremadamente
inestables con `qsec` como salida. Al convertirlas a dummies
en 2.2 se rompe esa linealidad forzada y los coeficientes
se estabilizan parcialmente, reduciendo el colapso en test.

## Conclusión
Aunque ninguno de los dos modelos generaliza bien, las dummies
son **claramente menos malas** con `qsec` como salida. El problema
de fondo sigue siendo el mismo: dataset demasiado pequeño para
OLS sin regularización.
