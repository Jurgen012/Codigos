```python
import pandas as pd
import numpy as np

df = pd.read_csv("Boston Housing Dataset 1978.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
CRIMescalable = (df["CRIM"] - df["CRIM"].min()) / (df["CRIM"].max() - df["CRIM"].min())
CRIMescalable.describe()
```

```python
#Hacer ciclo de escalar las otras variables (misma formula)

Escalas = pd.DataFrame()
Escalas["Variables"] = np.array(["CRIM","ZN", "INDUS", "CHAS", "NOX", "RM", "AGE", "DIS", "RAD", "TAX", "PTRATIO", "B", "LSTAT", "MEDV"])

DatasetEscalado = pd.DataFrame()
for i in Escalas["Variables"]:
    DatasetEscalado[i] = (df[i] - df[i].min()) / (df[i].max() - df[i].min())
DatasetEscalado.describe()
```

```python
Datasetmeanstd = pd.DataFrame()
for i in Escalas["Variables"]:
    DatasetEscalado[i] = (df[i] - df[i].mean()) / df[i].std() 
DatasetEscalado.describe()
```

```python
#ahora con escalamiento estandar 
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
for i in Escalas["Variables"]:
    DatasetEscalado[i] = scaler.fit_transform(df[[i]])
DatasetEscalado.describe()
```

```python
#histograma 
import matplotlib.pyplot as plt
DatasetEscalado.hist(bins=30, figsize=(15,10))
plt.show()
```
