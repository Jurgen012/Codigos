```python
import pandas as pd

agua = pd.read_excel("Aguas_datos.xlsx")
```

```python
agua.head()
```

```python
#quiero crear un nuevo dataframe con 0 o 1, dependiendo, por ejemplo en tamaño 1 es grande y 0 es chica, en hielo es 0 sin y 1 con, para sabor se crean dos columnas una para cada sabor, 0 si no esta 1 si si esta, y el score

df = pd.DataFrame()

df["Tamaño"] = agua["Tamano"].map ({"grande": 1, "chica": 0})
df["Hielo"] = agua["Hielo"].map ({"con": 1, "sin": 0})
df["kahlua"] = agua["Sabor"].map ({"kahlua": 1, "horchata": 0})
df["horchata"] = agua["Sabor"].map ({"horchata": 1, "kahlua": 0})    
df["Score"] = agua["Score"]

df.head()
```

```python
#parte 2, ahora quiero sin = -1, con = 1, grande = -1, chica = 1 
df2 = pd.DataFrame()
df2["Tamaño"] = agua["Tamano"].map ({"grande": -1, "chica": 1})
df2["Hielo"] = agua["Hielo"].map ({"con": 1, "sin": -1})
df2["kahlua"] = agua["Sabor"].map ({"kahlua": 1, "horchata": 0})
df2["horchata"] = agua["Sabor"].map ({"horchata": 1, "kahlua": 0})
df2["Score"] = agua["Score"]

df2.head()
```

```python
#parte 3 = comparar betas de la parte 1 con la parte 2, para ver si el signo de las betas cambia o no, y si el valor absoluto de las betas cambia o no
import statsmodels.api as sm
X = df[["Tamaño", "Hielo", "kahlua", "horchata"]]
y = df["Score"]
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```

```python
X = df2[["Tamaño", "Hielo", "kahlua", "horchata"]]
y = df2["Score"]
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```
